```python
#Task 1.1
def task1_1(input_value):
    input_value = str(input_value)
    if input_value.isdigit() == False:
        return -1
    if len(input_value) != 14:
        return -1
    else:
        pass
    arr = []
    for i in input_value:
        arr.append(i)
    for j in range(1,len(arr),2):
        arr[j] = int(arr[j])
        arr[j] = int(arr[j]) * 2
        arr[j] = str(arr[j])
    string = ''
    total = 0
    for k in arr:
        if len(k) == 2:
            string = str(k)
            k = int(string[0]) + int(string[1])
            total += k
        else:
            total += int(k)
    curr = total % 10
    if curr != 0:
        check_digit = 10 - curr
    else:
        check_digit = 0
    return check_digit
            
task1_1('14576567654934')
```

```python
#Task 1.2
print(task1_1('68931209548956') == 0)
print(task1_1(32189437540542) == 8)
print(task1_1('29842!B02A02I1') == -1)
print(task1_1(129754321468) == -1)
```

```python
#Task 1.3
def task1_3(input_value):
    input_value = str(input_value)
    if input_value.isdigit() == False or len(input_value) != 15:
        return False
    check_digit = input_value[14]
    input_value = input_value[:14]
    calculated = task1_1(input_value)
    check_digit = int(check_digit)
    if check_digit == calculated:
        return True
    else:
        return False

print(task1_3('68931209548956') == 0)
print(task1_3(32189437540542) == 8)
print(task1_3('29842!B02A02I1') == -1)
print(task1_3(129754321468) == -1)
```
