```python
#Task 4.2
import sqlite3
conn = sqlite3.connect('school.db')
from datetime import datetime
class Person:
    def __init__(self):
        self.full_name = None
        self.date_of_birth = None
    def is_adult(self):
        current_year = datetime.now().year
        year,month,day = self.date_of_birth.split('-')
        result = current_year - int(year)
        if result > 18:
            return True
        else:
            return False
    def screen_name(self):
        string = ''
        for i in self.full_name:
            if i.isalpha() == True:
                string += i
        year,month,day = self.date_of_birth.split('-')
        string += month
        string += day
        return string

p = Person()
p.full_name = "John Tan"
p.date_of_birth = "2000-06-01"
print(p.is_adult())
print(p.screen_name())

class Staff(Person):
    def screen_name(self):
        return super().screen_name() + "Staff"
    def is_adult(self):
        return True

class Student(Person):
    def is_adult(self):
        return False
    
with open('people.txt','r') as file:
    for line in file:
        name,dob,string = line.strip().split(',')
        if string == 'Staff':
            P = Staff()
            P.full_name = name
            P.date_of_birth = dob
        elif string == 'Student':
            P = Student()
            P.full_name = name
            P.date_of_birth = dob
        elif string =='Person':
            P = Person()
            P.full_name = name
            P.date_of_birth = dob
        conn.execute('''DROP TABLE IF EXISTS People''')   
        query =  ''' INSERT INTO People(FullName,DateOfBirth,ScreenName,IsAdult)
         VALUES(?,?,?,?)
         '''
        conn.execute(query,(P.full_name,P.date_of_birth,P.screen_name(),P.is_adult()))
    conn.commit()
    conn.close()
```

```python
#Task 4.3
from flask import *
app = Flask(__name__)
import sqlite3

@app.route('/', methods = ['GET','POST'])
def home():
    conn = sqlite3.connect('school.db')
    query = '''SELECT People.FullName, People.ScreenName, People.IsAdult
    FROM People'''
    cursor = conn.execute(query,()).fetchall()
    for doc in cursor:
        if doc[2] == 0:
            identity = "Student"
        else:
            identity = 'Staff'
    return render_template('home.html',cursor = cursor ,identity = identity)

if __name__ == '__main__':
    app.run(port = 8000)
```
