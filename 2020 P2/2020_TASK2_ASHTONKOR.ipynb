```python
#Task 2.1
from random import *
def task2_1(filename,quantity,maximum):
    with open(filename,'w') as file:
        for i in range(quantity):
            string = ''
            num = randint(0,maximum)
            string = str(num) + '\n'
            file.write(string)
task2_1('randomnumbers.txt',1000,5000)
```

```python
#Task 2.2
def task2_2(list_of_integers):
    def mergeSort(arr):
        n = len(arr)
        if n < 2:
            return arr
        left = mergeSort(arr[0:n // 2])
        right = mergeSort(arr[n // 2:n])
        return merge(left, right)
    
    def merge(left, right):
        results = [0] * (len(left) + len(right))
        i = 0
        j = 0
        index = 0
        while i < len(left) and j < len(right):
            if left[i] < right[j]:
                results[index] = left[i]
                i += 1
            else:
                results[index] = right[j]
                j += 1
            index += 1
        while i < len(left):
            results[index] = left[i]
            i += 1
            index += 1
        while j < len(right):
            results[index] = right[j]
            j += 1
            index += 1
        return results
    sorted_arr = mergeSort(list_of_integers)
    return sorted_arr

task2_2([56,25,4,98,0,18,4,5,7,0]) == [0, 0, 4, 4, 5, 7, 18, 25, 56, 98]
```

```python
#Task 2.3
def task2_3(filename_in,filename_out):
    with open(filename_in,'r') as file1:
        listt = []
        lines = file1.readlines()
        for line in lines:
            line = line.strip()
            listt.append(int(line))
        output = task2_2(listt)
    with open(filename_out,'w') as file2:
        file2.write(str(output))
        
task2_3('randomnumbers.txt','sortednumbers.txt')
```
