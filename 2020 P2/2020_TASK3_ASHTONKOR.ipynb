```python
#Task 3.1
def task3_1(quantity_of_data):
    if quantity_of_data[-2:] == 'KB' or quantity_of_data[-2:] == 'MB' or quantity_of_data[-2:] == 'GB' or quantity_of_data[-2:] == 'TB':
        print("Approved unit")
        if quantity_of_data[-2:] == 'KB':
            number = int(quantity_of_data[:-2]) * 10**3
            return number
        elif quantity_of_data[-2:] == 'MB':
            number = int(quantity_of_data[:-2]) * 10**6
            return number
        elif quantity_of_data[-2:] == 'GB':
            number = int(quantity_of_data[:-2]) * 10**9
            return number
        else:
            number = int(quantity_of_data[:-2]) * 10**12
            return number
    else:
        return "invalid data"

task3_1("8KB")
```

```python
#Task 3.2
def task3_2(quantity_of_data):
    if (quantity_of_data[-2:] == 'KB' or 
        quantity_of_data[-2:] == 'MB' or 
        quantity_of_data[-2:] == 'GB' or 
        quantity_of_data[-2:] == 'TB' or 
        quantity_of_data[-3:] == 'KiB' or 
        quantity_of_data[-3:] == 'MiB' or 
        quantity_of_data[-3:] == 'GiB' or 
        quantity_of_data[-3:] == 'TiB'):
        print('Approved unit')
        if quantity_of_data[-2:] == 'KB':
            number = int(quantity_of_data[:-2]) * 10**3
            return number
        elif quantity_of_data[-2:] == 'MB':
            number = int(quantity_of_data[:-2]) * 10**6
            return number
        elif quantity_of_data[-2:] == 'GB':
            number = int(quantity_of_data[:-2]) * 10**9
            return number
        elif quantity_of_data[-2:] == 'TB':
            number = int(quantity_of_data[:-2]) * 10**12
            return number
        elif quantity_of_data[-3:] == 'KiB':
            number = int(quantity_of_data[:-3]) * 2**10
            return number
        elif quantity_of_data[-3:] == 'MiB':
            number = int(quantity_of_data[:-3]) * 2**20
            return number
        elif quantity_of_data[-3:] == 'GiB':
            number = int(quantity_of_data[:-3]) * 2**30
            return number
        else:
            number = int(quantity_of_data[:-3]) * 2**40
            return number
    else:
        return 'invalid data'

task3_2('2MiB')
```
